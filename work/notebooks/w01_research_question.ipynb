```python
import pandas as pd
import numpy as np

# Creating mock dataset simulating 'dim_content' & 'fact_content_daily_performance'
np.random.seed(42)
n_rows = 1500

df = pd.DataFrame({
    'page_id': [f"page_{i}" for i in range(n_rows)],
    'impressions': np.random.randint(100, 5000, size=n_rows),
    'clicks': np.random.randint(5, 500, size=n_rows),
    'position': np.random.uniform(1.0, 50.0, size=n_rows),
    'word_count': np.random.randint(300, 2500, size=n_rows)
})

# Calculate required metrics for Section 3
total_pages = len(df)
avg_position = df['position'].mean()
avg_clicks = df['clicks'].mean()

print("--- Data Summary (Starter Data Analysis) ---")
print(f"1. Total Pages Analyzed: {total_pages:,}")
print(f"2. Average Search Position: {avg_position:.2f}")
print(f"3. Average Daily Clicks per Page: {avg_clicks:.2f}")
```
