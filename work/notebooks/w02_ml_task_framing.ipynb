## 1) My lane as an ML task

My lane is content refresh prioritization.

The ML task is a **binary classification** problem. The goal is to predict whether a web page is likely to be declining and should be considered for a content refresh.

The model would classify each page as either:
- declining
- not declining

The output can then be used to rank pages by their predicted likelihood of decline and prioritize which pages should be reviewed first.

## 2) Target or proxy

For the starter dataset, I can create a proxy target from the observed `trend_direction` field:

- `down` → 1 (declining)
- anything else → 0 (not declining)

This target represents the outcome that the model is trying to predict.

## 3) Success metric

The main success metric is **Precision@50**.

Precision@50 measures how many of the 50 highest-priority pages selected by the model are actually declining.

This metric is useful because the content team has limited time and cannot review every page. A good model should therefore put a high proportion of genuinely declining pages near the top of the refresh queue.

## 5) Why ML beats a fixed rule here

A fixed rule could select pages using simple thresholds, such as choosing pages that have not been updated for 180 days and have at least 500 impressions.

However, page decline can depend on several signals at the same time, including content age, days since update, impressions, average position, CTR, and word count.

A machine learning model can learn combinations and patterns across these signals instead of relying on one manually chosen threshold.

The model can produce a ranked list of pages, allowing the content team to focus its limited review time on the pages most likely to need a refresh.

ML is therefore useful here because the relationship between the available signals and page decline may be more complex than a single fixed rule.
